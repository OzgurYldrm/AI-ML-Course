```python
import numpy as np

def cosine_checker(function, tol=1e-6):
    vector_pairs = [
        # (v1, v2, true_similarity)
        (
            np.array([1, 0, 0]),
            np.array([1, 0, 0]),
            1.0
        ),
        (
            np.array([1, 0, 0]),
            np.array([0, 1, 0]),
            0.0
        ),
        (
            np.array([1, 1, 0]),
            np.array([1, 1, 0]),
            1.0
        ),
        (
            np.array([1, 2, 3]),
            np.array([4, 5, 6]),
            0.9746318461970762
        ),
        (
            np.array([-1, -1, 0]),
            np.array([1, 1, 0]),
            -1.0
        )
    ]

    score = 0

    for i, (v1, v2, true_sim) in enumerate(vector_pairs, 1):

        prediction = function(v1, v2)

        if np.isclose(prediction, true_sim, atol=tol):
            score += 1

        print(f"CASE {i}")
        print("V1:", v1)
        print("V2:", v2)
        print("TRUE:", true_sim)
        print("PREDICTION:", prediction)
        print("-" * 50)

    print(f"Score: {score}/5")

def tf_idf_checker(TFIDFClass, tol=1e-6):

    test_cases = [
        {
            "documents": [
                "apple banana apple",
                "banana orange",
                "banana banana apple"
            ],
            "expected_vocab": ["apple", "banana", "orange"],
            "expected_matrix": np.array([
                [0.27031007, 0.0,        0.0       ],
                [0.0,        0.0,        0.54930614],
                [0.13515504, 0.0,        0.0       ]
            ])
        },
        {
            "documents": [
                "cat dog",
                "dog mouse",
                "cat mouse mouse"
            ],
            "expected_vocab": ["cat", "dog", "mouse"],
            "expected_matrix": np.array([
                [0.20273255, 0.20273255, 0.0       ],
                [0.0,        0.20273255, 0.20273255],
                [0.13515504, 0.0,        0.27031007]
            ])
        }
    ]

    total_score = 0
    max_score = len(test_cases) * 4

    for case_id, case in enumerate(test_cases, 1):

        documents = case["documents"]
        expected_vocab = case["expected_vocab"]
        expected_matrix = case["expected_matrix"]

        model = TFIDFClass()
        model.fit(documents)

        student_matrix = model.transform(documents)
        student_fit_transform = model.fit_transform(documents)

        print("="*60)
        print(f"CASE {case_id}")
        print("="*60)

        # Vocab
        if sorted(model.vocab.keys()) == expected_vocab:
            print("Vocab: CORRECT")
            total_score += 1
        else:
            print("Vocab: WRONG")
            print("Expected:", expected_vocab)
            print("Got:", sorted(model.vocab.keys()))

        # Shape
        if student_matrix.shape == expected_matrix.shape:
            print("Matrix Shape: CORRECT")
            total_score += 1
        else:
            print("Matrix Shape: WRONG")

        # Transform
        if np.allclose(student_matrix, expected_matrix, atol=tol):
            print("Transform Values: CORRECT")
            total_score += 1
        else:
            print("Transform Values: WRONG")
            print("Expected:\n", expected_matrix)
            print("Got:\n", student_matrix)

        # Fit_Transform
        if np.allclose(student_fit_transform, expected_matrix, atol=tol):
            print("Fit_Transform: CORRECT")
            total_score += 1
        else:
            print("Fit_Transform: WRONG")

    print("\nFINAL SCORE:", f"{total_score}/{max_score}")

def self_attention_checker(SelfAttentionClass, tol=1e-6):
    Q = np.array([[1., 0.],
                  [0., 1.]])

    K = np.array([[1., 0.],
                  [0., 1.]])

    V = np.array([[1., 2.],
                  [3., 4.]])

    expected_output = np.array([
        [1.6604769 , 2.6604769 ],
        [2.3395231 , 3.3395231 ]
    ])

    model = SelfAttentionClass()
    student_output = model.forward(Q, K, V)

    print("="*50)
    print("SELF ATTENTION TEST")
    print("="*50)

    # Shape test
    if student_output.shape == expected_output.shape:
        print("Shape: CORRECT")
        shape_score = 1
    else:
        print("Shape: WRONG")
        shape_score = 0

    # Value test
    if np.allclose(student_output, expected_output, atol=tol):
        print("Values: CORRECT")
        value_score = 1
    else:
        print("Values: WRONG")
        print("Expected:\n", expected_output)
        print("Got:\n", student_output)
        value_score = 0

    total = shape_score + value_score
    print("\nFINAL SCORE:", f"{total}/2")
```
